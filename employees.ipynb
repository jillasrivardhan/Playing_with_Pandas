```python
def who_done_most_projects(num):
   if num >=1 & num <= 5:
       return "Low"
   elif num >=6 & num <= 10:
       return "Medium"
   elif num >=11 & num <= 15:
       return "High" 
   else:
       return "Very High"
```

```python
df['status'] = df['projects_completed'].apply(who_done_most_projects)
```

```python
df.head()
```

```python
def bonus(df):
   if df['status'] == "very high":
       return df['salary'] * 1000
   else:
       return df['salary'] * 500
```
